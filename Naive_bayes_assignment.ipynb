```python
import numpy as np
import pandas as pd

data = {
    "Message": [
        "free money",
        "win prize",
        "free prize",
        "free cash",
        "win money",
        "free offer",
        "claim prize",
        "free reward",
        "meeting today",
        "project meeting"
    ],

    "Class": [
        "Spam",
        "Spam",
        "Spam",
        "Spam",
        "Spam",
        "Spam",
        "Spam",
        "Spam",
        "Not Spam",
        "Not Spam"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
spam_count = np.sum(df["Class"] == "Spam")
not_spam_count = np.sum(df["Class"] == "Not Spam")
#total
total_count = len(df)
p_spam = spam_count/total_count
p_not_spam = not_spam_count/total_count
print(p_spam)
print(p_not_spam)
```

```python
def word_probablity(word,class_name):
  messages = df[df["Class"] ==class_name]["Message"]
  count = 0
  for message in messages:
    words = message.lower().split()
    if word in words:
      count +=1
    class_count = len(messages)
    probablity = (count+1)/(class_count +2)
  return probablity
print(word_probablity("meeting", "Not Spam"))
```

```python
word = "meeting"
p_spam_free = word_probablity(word,"Spam")
p_not_spam_free = word_probablity(word,"Not Spam")
print(p_spam_free)
print(p_not_spam_free)
spam_score = p_spam_free * p_spam
not_spam_score = p_not_spam_free * p_not_spam
print("spam score",spam_score)
print("not spam score",not_spam_score)
```

```python
if spam_score > not_spam_score:
  prediction ="Spam"
  print("spam")
else:
  prediction="Not spam"
  print("not spam")
print("prediction",prediction)
```
